```python
from mupt.reactions.polymerizeit.make_pi import make_pi

inputs = make_pi(inputs_file="epoxy_det-epo.yaml")

print('\n\nMolecules:')
print([inputs['molecules'][i]['name'] for i in range(len(inputs['molecules']))])
print('Monomers:')
print([inputs['monomers'][i]['name'] for i in range(len(inputs['monomers']))])
print('Repeat Units:')
print([inputs['repeat_units'][i]['name'] for i in range(len(inputs['repeat_units']))])
print('Byproducts:')
print([inputs['byproducts'][i]['name'] for i in range(len(inputs['byproducts']))])
```

```python
import rdkit

for j, molecule in enumerate(inputs['molecules']):
    if 'mol' in molecule:
        print(f"Drawing molecule {molecule['name']}:")
        print(f"SMILES: {rdkit.Chem.MolToSmiles(rdkit.Chem.MolFromSmiles(molecule['smi']))}")
        display(rdkit.Chem.Draw.MolToImage(rdkit.Chem.MolFromSmiles(molecule['smi'])))
```
