```python
from mupt.reactions.polymerizeit.make_pi import make_pi

inputs = make_pi(inputs_file="inputs.yaml")
```

```python
import rdkit

for j, molecule in enumerate(inputs['monomers']):
    if 'mol' in molecule:
        print(f"Drawing molecule {molecule['name']}:")
        print(f"SMILES: {rdkit.Chem.MolToSmiles(rdkit.Chem.MolFromSmiles(molecule['smi']))}")
        display(rdkit.Chem.Draw.MolToImage(molecule['mol']))
```
